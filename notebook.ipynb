## Preprocessing Pipeline

Steps:
1. **Drop** the `id` column (row index, not a feature)
2. **Impute** `Arrival Delay in Minutes` (292 nulls) with the median
3. **Cap outliers** on delay columns using IQR upper fence (99th percentile cap)
4. **Encode target**: `satisfaction` True/False → 1/0
5. **Label-encode** categoricals for a fully numeric version (`df_processed`)
6. **Train/test split** — stratified 80/20

```python
# ── 1 & 2: Drop id, median-impute all numeric columns ──────────────────────
from sklearn.impute import SimpleImputer

df_clean = df

# Identify numeric columns (the only NaN column is Arrival Delay, but we fit
# on ALL numerics so the imputer can be reused on test.csv safely)
numeric_cols_impute = df_clean.select_dtypes(include='number').columns.tolist()
# Remove the target — we never impute it
numeric_cols_impute = [c for c in numeric_cols_impute if c != 'satisfaction']

imputer = SimpleImputer(strategy='median')
df_clean[numeric_cols_impute] = imputer.fit_transform(df_clean[numeric_cols_impute])

# Keep arrival_median for backward compatibility with downstream cells
arrival_median = imputer.statistics_[
    numeric_cols_impute.index('Arrival Delay in Minutes')
]

print(f'Numeric columns imputed: {len(numeric_cols_impute)}')
print(f'Arrival Delay median used: {arrival_median}')
print(f'Missing values after imputation: {df_clean.isnull().sum().sum()}')
print(f'Shape: {df_clean.shape}')
```

```python
# ── 3: Cap outliers on skewed delay columns (IQR is 0, use 99th percentile) ─
delay_cols = ['Departure Delay in Minutes', 'Arrival Delay in Minutes']

caps = {}
for col in delay_cols:
    cap = df_clean[col].quantile(0.99)
    caps[col] = cap
    df_clean[col] = df_clean[col].clip(upper=cap)

print('99th-percentile caps applied:')
for col, cap in caps.items():
    print(f'  {col}: {cap:.1f} min')
```

```python
# ── 4: Encode target ────────────────────────────────────────────────────────
df_clean['satisfaction'] = df_clean['satisfaction'].astype(int)
print('Target distribution:')
print(df_clean['satisfaction'].value_counts().rename({0: 'Not Satisfied', 1: 'Satisfied'}))
```

```python
# ── 5: Label-encode categoricals → fully numeric df_processed ───────────────
from sklearn.preprocessing import LabelEncoder

cat_cols = ['Gender', 'Customer Type', 'Type of Travel', 'Class']
df_processed = df_clean.copy()

label_encoders = {}
for col in cat_cols:
    le = LabelEncoder()
    df_processed[col] = le.fit_transform(df_clean[col])
    label_encoders[col] = le
    print(f'{col}: {dict(zip(le.classes_, le.transform(le.classes_)))}')

print(f'\ndf_processed dtypes:\n{df_processed.dtypes}')
```

```python
# ── 6: Train / test split ───────────────────────────────────────────────────
from sklearn.model_selection import train_test_split

X_proc = df_processed.drop(columns=['satisfaction'])
y_proc = df_processed['satisfaction']

X_proc_train, X_proc_test, y_proc_train, y_proc_test = train_test_split(
    X_proc, y_proc, test_size=0.2, random_state=42, stratify=y_proc
)

print(f'Train : {X_proc_train.shape[0]:>7,} rows | {X_proc_train.shape[1]} features')
print(f'Test  : {X_proc_test.shape[0]:>7,} rows | {X_proc_test.shape[1]} features')
print(f'\nClass balance (train):')
print((y_proc_train.value_counts(normalize=True) * 100).rename({0: 'Not Satisfied %', 1: 'Satisfied %'}).round(2))
```

## Feature Engineering

New features built from related columns:

**Delay features** (Departure Delay + Arrival Delay)
- `total_delay` — combined delay burden felt by the passenger
- `delay_diff` — arrival minus departure delay (negative = airline recovered time, positive = got worse)
- `delay_recovered` — flag: airline recovered ≥1 min in the air
- `any_delay` — binary flag: was there any delay at all?

**Service composite scores** (rating columns grouped by theme)
- `comfort_score` — Seat comfort + Leg room service
- `digital_score` — Inflight wifi + Online boarding + Ease of Online booking
- `inflight_exp_score` — Inflight entertainment + Food and drink + Cleanliness + On-board service
- `ground_score` — Gate location + Checkin service + Baggage handling

**Interaction features**
- `service_range` — max rating − min rating across all service columns (consistency of experience)
- `avg_service_rating` — mean of all 14 service rating columns
- `dist_per_age` — Flight Distance / Age (long-haul relative to passenger age)

```python
# ── Delay features ──────────────────────────────────────────────────────────
df_feat = df_processed.copy()

dep = df_feat['Departure Delay in Minutes']
arr = df_feat['Arrival Delay in Minutes']

df_feat['total_delay']      = dep + arr
df_feat['delay_diff']       = arr - dep          # >0 got worse, <0 recovered
df_feat['delay_recovered']  = (df_feat['delay_diff'] < 0).astype(int)
df_feat['any_delay']        = ((dep > 0) | (arr > 0)).astype(int)

print('Delay features added:')
df_feat[['total_delay', 'delay_diff', 'delay_recovered', 'any_delay']].describe().round(2)
```

```python
# ── Service composite scores ─────────────────────────────────────────────────
comfort_cols  = ['Seat comfort', 'Leg room service']
digital_cols  = ['Inflight wifi service', 'Online boarding', 'Ease of Online booking']
inflight_cols = ['Inflight entertainment', 'Food and drink', 'Cleanliness', 'On-board service']
ground_cols   = ['Gate location', 'Checkin service', 'Baggage handling']

df_feat['comfort_score']     = df_feat[comfort_cols].mean(axis=1)
df_feat['digital_score']     = df_feat[digital_cols].mean(axis=1)
df_feat['inflight_exp_score']= df_feat[inflight_cols].mean(axis=1)
df_feat['ground_score']      = df_feat[ground_cols].mean(axis=1)

print('Service composite scores added:')
df_feat[['comfort_score', 'digital_score', 'inflight_exp_score', 'ground_score']].describe().round(2)
```

```python
# ── Interaction features ─────────────────────────────────────────────────────
all_service_cols = comfort_cols + digital_cols + inflight_cols + ground_cols + \
                  ['Departure/Arrival time convenient']

df_feat['service_range']      = df_feat[all_service_cols].max(axis=1) - df_feat[all_service_cols].min(axis=1)
df_feat['avg_service_rating'] = df_feat[all_service_cols].mean(axis=1)
df_feat['dist_per_age']       = df_feat['Flight Distance'] / df_feat['Age']

print('Interaction features added:')
df_feat[['service_range', 'avg_service_rating', 'dist_per_age']].describe().round(2)
```

```python
# ── Summary: new feature columns and final shape ─────────────────────────────
new_cols = ['total_delay', 'delay_diff', 'delay_recovered', 'any_delay',
            'comfort_score', 'digital_score', 'inflight_exp_score', 'ground_score',
            'service_range', 'avg_service_rating', 'dist_per_age']

print(f'Original features : {df_processed.shape[1] - 1}')
print(f'Engineered features added: {len(new_cols)}')
print(f'Total features in df_feat : {df_feat.shape[1] - 1}')
print()

# Correlation of new features with target
corr = df_feat[new_cols + ['satisfaction']].corr()['satisfaction'].drop('satisfaction').sort_values(key=abs, ascending=False)
print('Correlation of engineered features with satisfaction:')
print(corr.round(4))
```

```python
# ── Retrain CatBoost on engineered features ──────────────────────────────────
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
from catboost import CatBoostClassifier

X_fe = df_feat.drop(columns=['satisfaction'])
y_fe = df_feat['satisfaction']

X_fe_train, X_fe_test, y_fe_train, y_fe_test = train_test_split(
    X_fe, y_fe, test_size=0.2, random_state=42, stratify=y_fe
)

# Categorical columns are now label-encoded integers — pass as cat_features
cat_features_fe = ['Gender', 'Customer Type', 'Type of Travel', 'Class']

model_fe = CatBoostClassifier(
    iterations=500,
    learning_rate=0.1,
    depth=6,
    eval_metric='Accuracy',
    cat_features=cat_features_fe,
    random_seed=42,
    verbose=100
)

model_fe.fit(
    X_fe_train, y_fe_train,
    eval_set=(X_fe_test, y_fe_test),
    early_stopping_rounds=30
)

y_fe_pred = model_fe.predict(X_fe_test)
print(f'\nAccuracy (with feature engineering): {accuracy_score(y_fe_test, y_fe_pred):.4f}')
print()
print(classification_report(y_fe_test, y_fe_pred, target_names=['Not Satisfied', 'Satisfied']))
```

## Outlier Analysis (IQR method)

```python
# Only continuous columns are meaningful for outlier detection;
# rating columns (0-5) are bounded ordinal scales — skip them.
continuous_cols = ['Age', 'Flight Distance', 'Departure Delay in Minutes', 'Arrival Delay in Minutes']

outlier_summary = []
for col in continuous_cols:
    series = df[col].dropna()
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    n_outliers = ((series < lower) | (series > upper)).sum()
    pct = n_outliers / len(series) * 100
    outlier_summary.append({
        'Column': col,
        'Q1': Q1, 'Q3': Q3, 'IQR': IQR,
        'Lower Bound': lower, 'Upper Bound': upper,
        'Max Value': series.max(),
        '# Outliers': n_outliers,
        '% Outliers': round(pct, 2)
    })

outlier_df = pd.DataFrame(outlier_summary).set_index('Column')
outlier_df
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 4, figsize=(18, 5))
for ax, col in zip(axes, continuous_cols):
    ax.boxplot(df[col].dropna(), vert=True, patch_artist=True,
               boxprops=dict(facecolor='#4c72b0', alpha=0.6),
               medianprops=dict(color='red', linewidth=2),
               flierprops=dict(marker='o', markersize=2, alpha=0.3, color='orange'))
    ax.set_title(col, fontsize=10)
    ax.set_xticks([])

fig.suptitle('Boxplots — Continuous Features (outliers shown in orange)', fontsize=12)
plt.tight_layout()
plt.show()
```

## K-Fold Cross-Validation

Evaluate the CatBoost model (with feature engineering) using **5-fold Stratified K-Fold** cross-validation on the full dataset.

- `StratifiedKFold` preserves the class balance (~55 / 45) in every fold.
- Each fold trains on 4/5 of the data (~559 K rows) and validates on the remaining 1/5 (~140 K rows).
- Metrics reported per fold: **Accuracy** and **ROC-AUC**.
- Early stopping is applied inside each fold using a small internal hold-out (10 % of the fold's training data) to avoid overfitting without leaking the validation fold.

```python
import numpy as np
import pandas as pd
from catboost import CatBoostClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, roc_auc_score

# ── Data: use the fully engineered feature matrix ────────────────────────────
X_cv = df_feat.drop(columns=['satisfaction'])
y_cv = df_feat['satisfaction']

cat_features_cv = ['Gender', 'Customer Type', 'Type of Travel', 'Class']

N_SPLITS = 5
skf = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=42)

fold_results = []

for fold, (train_idx, val_idx) in enumerate(skf.split(X_cv, y_cv), start=1):
    X_fold_train, X_fold_val = X_cv.iloc[train_idx], X_cv.iloc[val_idx]
    y_fold_train, y_fold_val = y_cv.iloc[train_idx], y_cv.iloc[val_idx]

    # Small internal hold-out (10 %) for early stopping — prevents leaking val fold
    n_internal = int(len(X_fold_train) * 0.1)
    X_es, y_es = X_fold_train.iloc[-n_internal:], y_fold_train.iloc[-n_internal:]
    X_tr, y_tr = X_fold_train.iloc[:-n_internal], y_fold_train.iloc[:-n_internal]

    model_cv = CatBoostClassifier(
        iterations=500,
        learning_rate=0.1,
        depth=6,
        eval_metric='Accuracy',
        cat_features=cat_features_cv,
        random_seed=42,
        verbose=0,
    )

    model_cv.fit(
        X_tr, y_tr,
        eval_set=(X_es, y_es),
        early_stopping_rounds=30,
    )

    y_pred_cv  = model_cv.predict(X_fold_val)
    y_proba_cv = model_cv.predict_proba(X_fold_val)[:, 1]

    acc     = accuracy_score(y_fold_val, y_pred_cv)
    roc_auc = roc_auc_score(y_fold_val, y_proba_cv)

    fold_results.append({
        'Fold'          : fold,
        'Best Iteration': model_cv.best_iteration_,
        'Accuracy'      : round(acc, 4),
        'ROC-AUC'       : round(roc_auc, 4),
    })
    print(f'Fold {fold}/{N_SPLITS}  |  Accuracy: {acc:.4f}  |  ROC-AUC: {roc_auc:.4f}  |  Best iter: {model_cv.best_iteration_}')

cv_df = pd.DataFrame(fold_results)

# Summary row
summary = pd.DataFrame([{
    'Fold'          : 'Mean +/- Std',
    'Best Iteration': f"{cv_df['Best Iteration'].mean():.1f} +/- {cv_df['Best Iteration'].std():.1f}",
    'Accuracy'      : f"{cv_df['Accuracy'].mean():.4f} +/- {cv_df['Accuracy'].std():.4f}",
    'ROC-AUC'       : f"{cv_df['ROC-AUC'].mean():.4f} +/- {cv_df['ROC-AUC'].std():.4f}",
}])

display(pd.concat([cv_df.astype(str), summary], ignore_index=True))
```

## K-Fold Cross-Validation — CatBoost with L2 Regularisation

Repeat the 5-fold Stratified K-Fold experiment adding **L2 leaf regularisation** (`l2_leaf_reg`) to CatBoost.

`l2_leaf_reg` penalises large leaf weights, acting as L2 (ridge) regularisation on the tree leaves. A higher value shrinks the leaf scores towards zero and reduces overfitting at the cost of some bias.

Three values are tested: **1** (CatBoost default), **5**, and **10**.  
The fold loop is identical to the baseline CV above, so results are directly comparable.


```python
import numpy as np
import pandas as pd
from catboost import CatBoostClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, roc_auc_score

cat_features_cv = ['Gender', 'Customer Type', 'Type of Travel', 'Class']

X_cv = df_feat.drop(columns=['satisfaction'])
y_cv = df_feat['satisfaction']

N_SPLITS  = 5
L2_VALUES = [1, 5, 10]   # l2_leaf_reg values to compare

skf = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=42)

all_results = []

for l2 in L2_VALUES:
    print(f'\n--- l2_leaf_reg = {l2} ---')
    fold_results = []

    for fold, (train_idx, val_idx) in enumerate(skf.split(X_cv, y_cv), start=1):
        X_fold_train, X_fold_val = X_cv.iloc[train_idx], X_cv.iloc[val_idx]
        y_fold_train, y_fold_val = y_cv.iloc[train_idx], y_cv.iloc[val_idx]

        # 10 % internal hold-out for early stopping (no leakage of the validation fold)
        n_internal = int(len(X_fold_train) * 0.1)
        X_es, y_es = X_fold_train.iloc[-n_internal:], y_fold_train.iloc[-n_internal:]
        X_tr, y_tr = X_fold_train.iloc[:-n_internal], y_fold_train.iloc[:-n_internal]

        model_l2 = CatBoostClassifier(
            iterations=500,
            learning_rate=0.1,
            depth=6,
            l2_leaf_reg=l2,
            eval_metric='Accuracy',
            cat_features=cat_features_cv,
            random_seed=42,
            verbose=0,
        )

        model_l2.fit(
            X_tr, y_tr,
            eval_set=(X_es, y_es),
            early_stopping_rounds=30,
        )

        y_pred_l2  = model_l2.predict(X_fold_val)
        y_proba_l2 = model_l2.predict_proba(X_fold_val)[:, 1]

        acc     = accuracy_score(y_fold_val, y_pred_l2)
        roc_auc = roc_auc_score(y_fold_val, y_proba_l2)

        fold_results.append({
            'l2_leaf_reg'   : l2,
            'Fold'          : fold,
            'Best Iteration': model_l2.best_iteration_,
            'Accuracy'      : round(acc, 4),
            'ROC-AUC'       : round(roc_auc, 4),
        })
        print(f'  Fold {fold}/{N_SPLITS}  |  Accuracy: {acc:.4f}  |  ROC-AUC: {roc_auc:.4f}  |  Best iter: {model_l2.best_iteration_}')

    all_results.extend(fold_results)

all_df = pd.DataFrame(all_results)

# ── Summary table: mean +/- std per l2 value ─────────────────────────────────
summary_rows = []
for l2, grp in all_df.groupby('l2_leaf_reg'):
    summary_rows.append({
        'l2_leaf_reg'   : l2,
        'Accuracy'      : f"{grp['Accuracy'].mean():.4f} +/- {grp['Accuracy'].std():.4f}",
        'ROC-AUC'       : f"{grp['ROC-AUC'].mean():.4f} +/- {grp['ROC-AUC'].std():.4f}",
        'Best Iteration': f"{grp['Best Iteration'].mean():.1f} +/- {grp['Best Iteration'].std():.1f}",
    })

summary_df = pd.DataFrame(summary_rows).set_index('l2_leaf_reg')
print('\n=== 5-Fold CV Summary by l2_leaf_reg ===')
display(summary_df)
```

## Predict on Test Set — CatBoost K-Fold Ensemble

5-fold `StratifiedKFold` ensemble using the high-performance CatBoost config (`iterations=3000`, `learning_rate=0.05`, `depth=8`, `eval_metric=AUC`).

**Key differences vs. earlier runs:**
- Raw string categoricals are passed directly to CatBoost (no label-encoding) — CatBoost's internal ordered target statistics handle them more accurately.
- `early_stopping_rounds=150` gives the low learning-rate model enough patience to find the true optimum.
- Each fold predicts on the full test set; the 5 probability arrays are **averaged** for the final submission.
- OOF probabilities are also collected for a clean train-set AUC without touching the test file.

```python
import pandas as pd
import numpy as np
from catboost import CatBoostClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score

# ── Build feature matrix with RAW string categoricals (from df_clean) ─────────
# df_clean has already had: id dropped, Arrival Delay imputed, delays capped,
# satisfaction encoded to int — but cats are still raw strings.
cat_cols_str = ['Gender', 'Customer Type', 'Type of Travel', 'Class']

df_feat_raw = df_clean.copy()   # satisfaction already int; cats still strings

dep = df_feat_raw['Departure Delay in Minutes']
arr = df_feat_raw['Arrival Delay in Minutes']
df_feat_raw['total_delay']       = dep + arr
df_feat_raw['delay_diff']        = arr - dep
df_feat_raw['delay_recovered']   = (df_feat_raw['delay_diff'] < 0).astype(int)
df_feat_raw['any_delay']         = ((dep > 0) | (arr > 0)).astype(int)

comfort_cols  = ['Seat comfort', 'Leg room service']
digital_cols  = ['Inflight wifi service', 'Online boarding', 'Ease of Online booking']
inflight_cols = ['Inflight entertainment', 'Food and drink', 'Cleanliness', 'On-board service']
ground_cols   = ['Gate location', 'Checkin service', 'Baggage handling']

df_feat_raw['comfort_score']      = df_feat_raw[comfort_cols].mean(axis=1)
df_feat_raw['digital_score']      = df_feat_raw[digital_cols].mean(axis=1)
df_feat_raw['inflight_exp_score'] = df_feat_raw[inflight_cols].mean(axis=1)
df_feat_raw['ground_score']       = df_feat_raw[ground_cols].mean(axis=1)

all_service_cols = comfort_cols + digital_cols + inflight_cols + ground_cols + \
                   ['Departure/Arrival time convenient']
df_feat_raw['service_range']      = df_feat_raw[all_service_cols].max(axis=1) - df_feat_raw[all_service_cols].min(axis=1)
df_feat_raw['avg_service_rating'] = df_feat_raw[all_service_cols].mean(axis=1)
df_feat_raw['dist_per_age']       = df_feat_raw['Flight Distance'] / df_feat_raw['Age']

X_full = df_feat_raw.drop(columns=['satisfaction'])
y_full = df_feat_raw['satisfaction']

# Fill any NaN in cat cols with a sentinel (CatBoost requires no NaN in cat cols)
for col in cat_cols_str:
    X_full[col] = X_full[col].fillna('__MISSING__').astype(str)

# ── Preprocess test.csv with the same pipeline ────────────────────────────────
df_test_raw = pd.read_csv('test.csv')
test_ids    = df_test_raw['id'].copy()
df_test     = df_test_raw.drop(columns=['id']).copy()

# Median-impute all numeric columns using the imputer fitted on training data
df_test[numeric_cols_impute] = imputer.transform(df_test[numeric_cols_impute])
for col, cap in caps.items():
    df_test[col] = df_test[col].clip(upper=cap)

# Keep categoricals as raw strings
for col in cat_cols_str:
    df_test[col] = df_test[col].fillna('__MISSING__').astype(str)

dep_t = df_test['Departure Delay in Minutes']
arr_t = df_test['Arrival Delay in Minutes']
df_test['total_delay']       = dep_t + arr_t
df_test['delay_diff']        = arr_t - dep_t
df_test['delay_recovered']   = (df_test['delay_diff'] < 0).astype(int)
df_test['any_delay']         = ((dep_t > 0) | (arr_t > 0)).astype(int)

df_test['comfort_score']      = df_test[comfort_cols].mean(axis=1)
df_test['digital_score']      = df_test[digital_cols].mean(axis=1)
df_test['inflight_exp_score'] = df_test[inflight_cols].mean(axis=1)
df_test['ground_score']       = df_test[ground_cols].mean(axis=1)
df_test['service_range']      = df_test[all_service_cols].max(axis=1) - df_test[all_service_cols].min(axis=1)
df_test['avg_service_rating'] = df_test[all_service_cols].mean(axis=1)
df_test['dist_per_age']       = df_test['Flight Distance'] / df_test['Age']

X_test_final = df_test[X_full.columns]

# ── 5-Fold K-Fold ensemble ─────────────────────────────────────────────────────
N_SPLITS = 5
skf = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=42)

test_probas = np.zeros(len(X_test_final))
oof_probas  = np.zeros(len(X_full))

for fold, (train_idx, val_idx) in enumerate(skf.split(X_full, y_full), start=1):
    X_tr, X_val = X_full.iloc[train_idx], X_full.iloc[val_idx]
    y_tr, y_val = y_full.iloc[train_idx],  y_full.iloc[val_idx]

    # 10 % internal hold-out for early stopping — validation fold stays clean
    n_es = int(len(X_tr) * 0.1)
    X_es, y_es = X_tr.iloc[-n_es:], y_tr.iloc[-n_es:]
    X_tr, y_tr = X_tr.iloc[:-n_es],  y_tr.iloc[:-n_es]

    model_fold = CatBoostClassifier(
        iterations=3000,
        cat_features=cat_cols_str,
        learning_rate=0.05,
        depth=8,
        loss_function='Logloss',
        eval_metric='AUC',
        random_seed=42,
        verbose=200,
        allow_writing_files=False,
        thread_count=-1,
    )
    model_fold.fit(X_tr, y_tr, eval_set=(X_es, y_es), early_stopping_rounds=150)

    oof_probas[val_idx]  = model_fold.predict_proba(X_val)[:, 1]
    test_probas         += model_fold.predict_proba(X_test_final)[:, 1]
    oof_auc = roc_auc_score(y_val, oof_probas[val_idx])
    print(f'Fold {fold}/{N_SPLITS}  |  best iter: {model_fold.best_iteration_}  |  OOF AUC: {oof_auc:.5f}')

test_probas /= N_SPLITS

overall_oof_auc = roc_auc_score(y_full, oof_probas)
print(f'\nOverall OOF AUC: {overall_oof_auc:.5f}')

# ── Save result ────────────────────────────────────────────────────────────────
result_with_cat_boost = pd.DataFrame({
    'id'          : test_ids.values,
    'satisfaction': test_probas,
})
result_with_cat_boost.to_csv('result_with_cat_boost.csv', index=False)

print(f'Saved result_with_cat_boost.csv  ({len(result_with_cat_boost):,} rows)')
display(result_with_cat_boost.head(10))
```
